```python
import spacy
import re

# Load a spaCy language model
nlp = spacy.load("en_core_web_sm")

text = "Apple is looking at buying U.K. startup for $1 billion. This is an interesting development."
doc = nlp(text)

print("\n--- spaCy Parsing (Dependencies) ---")
for token in doc:
    print(f"{token.text} -> {token.dep_} -> {token.head.text}")
```

```python
print("\n--- spaCy Chunking (Noun Chunks) ---")
for chunk in doc.noun_chunks:
    print(f"{chunk.text} -> {chunk.label_} -> {chunk.root.text}")
```

```python
print("\n--- Custom Chunking with spaCy Matcher (e.g., Verb Phrases) ---")
from spacy.matcher import Matcher
matcher = Matcher(nlp.vocab)

# Example pattern for a simple verb phrase: VERB + (DET/ADJ/NOUN)* + NOUN
pattern = [{"POS": "VERB"}, {"POS": {"IN": ["DET", "ADJ", "NOUN"]}, "OP": "*"}, {"POS": "NOUN"}]
matcher.add("VERB_PHRASE", [pattern])

matches = matcher(doc)
for match_id, start, end in matches:
    span = doc[start:end]
    print(f"Found verb phrase: {span.text}")
```

```python
print("\n--- RegEx for simple pattern matching (e.g., dollar amounts) ---")
# Find patterns like '$1 billion'
regex_pattern = r"\$[0-9]+\s\w+"
found_patterns = re.findall(regex_pattern, text)
print(f"Found with RegEx: {found_patterns}")
```
