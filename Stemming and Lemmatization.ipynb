### Stemming and Lemmatization

**Stemming** is the process of reducing inflected (or sometimes derived) words to their word stem, base or root form—generally a written word form. The stem is not necessarily identical to the morphological root of the word; it is usually an affix-stripped version of the word. For example, 'consultant', 'consulting', 'consultative' might all be reduced to 'consult'.

**Lemmatization** is the process of grouping together the inflected forms of a word so they can be analysed as a single item, identified by the word's lemma, or dictionary form. For instance, 'consultant', 'consulting', 'consultative' and 'consult' would all map to 'consult', which is the base form of the word.

```python
# Re-executing the cell to define the sample text
sample_text = "Running, runs, runner, and ran are all forms of the verb run. Similarly, dogs, running, played, and playing are interesting words."
print(f"Original Text: {sample_text}\n")
```

```python
# Re-executing the cell to perform tokenization, stemming, and lemmatization
words = word_tokenize(sample_text)

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Perform Stemming
stemmed_words = [stemmer.stem(word) for word in words]
print(f"Stemmed Words: {stemmed_words}\n")

# Perform Lemmatization
lemmatized_words_default = [lemmatizer.lemmatize(word) for word in words]
print(f"Lemmatized Words (default): {lemmatized_words_default}\n")

lemmatized_words_verb = [lemmatizer.lemmatize(word, pos='v') for word in words]
print(f"Lemmatized Words (verb pos): {lemmatized_words_verb}")
```
