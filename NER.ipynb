Now that spaCy is installed and we have a model, let's import the library and load the model. Then we can process some sample text to extract named entities.

```python
import spacy

# Load the pre-trained English model
nlp = spacy.load("en_core_web_sm")

# Sample text for NER
text = "Apple Inc. was founded by Steve Jobs, Steve Wozniak, and Ronald Wayne on April 1, 1976, in Cupertino, California. Its headquarters are located at 1 Apple Park Way, Cupertino."

# Process the text with the spaCy model
doc = nlp(text)

print("Named Entities and their labels:")
for ent in doc.ents:
    print(f"Text: {ent.text}, Label: {ent.label_}")
```

The code above loads the spaCy model, processes a sample text, and then prints all identified named entities along with their respective labels (e.g., PERSON, ORG, GPE, DATE).
